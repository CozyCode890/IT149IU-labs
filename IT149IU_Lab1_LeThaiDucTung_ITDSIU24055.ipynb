### Exercise 3 – Arithmetic operators

```python
left_operands = [-5, 0, 5, 7.5]
right_operand = 2
```

```python
for a in left_operands:
    print(f"\nTest with left operand = {a}")
    print(a, "+", right_operand, "=", a + right_operand)
    print(a, "-", right_operand, "=", a - right_operand)
    print(a, "*", right_operand, "=", a * right_operand)
    print(a, "/", right_operand, "=", a / right_operand)
    print(a, "//", right_operand, "=", a // right_operand)
    print(a, "%", right_operand, "=", a % right_operand)
    print(a, "**", right_operand, "=", a ** right_operand)
```

### Exercise 4 – Eggs in boxes

```python
eggs = 28
per_box = 6

print("Total eggs:", eggs)
```

```python
# In conditon, modulo returns true if result is non-zero.
boxes = eggs // per_box + (1 if eggs % per_box else 0)
print("Boxes needed:", boxes)
```

```python
last_box_fill = eggs % per_box if eggs % per_box else per_box
print("Eggs in last box:", last_box_fill)
```

```python
need_to_fill_last = per_box - (eggs % per_box) if eggs % per_box else 0
print("Eggs needed to fill last box:", need_to_fill_last)
```

#### Exercise 8 – Name score with ord()

```python
def name_score(name):
    return sum(ord(c) for c in name)
```

```python
tom_score = name_score("Tom")
print("Tom score:", tom_score)
```

```python
jim_score = name_score("Jim")
print("Jim score:", jim_score)
```

```python
if tom_score > jim_score:
    print("Tom goes first!")
elif jim_score > tom_score:
    print("Jim goes first!")
else:
    print("It's a tie!")
```

#### Exercise 9 – Course grades (average, best, worst)

```python
courses = {}
```

```python
courses['Math'] = int(input('Enter grade for Math: '))
courses['Physics'] = int(input('Enter grade for Physics: '))
courses['AI'] = int(input('Enter grade for AI: '))
```

```python
avg = sum(courses.values()) / len(courses)
print('Average grade:', round(avg, 2))
```

```python
best = max(courses, key=courses.get)
print('Highest grade:', best, courses[best])
```

```python
worst = min(courses, key=courses.get)
print('Lowest grade:', worst, courses[worst])
```

#### Exercise 13 – BMI calculator

```python
weight = float(input("Enter your weight:"))
height = float(input("Enter your height:"))
print("Weight (kg):", weight)
print("Height (m):", height)
```

```python
BMI = weight / (height ** 2)
print("BMI =", BMI, "-> ", end="")

if BMI < 18.5:
    print("Underweight")
elif BMI < 25:
    print("Normal")
elif BMI < 30:
    print("Overweight")
else:
    print("Obese")
```

#### Exercise 14 – Cash withdrawal (//, %)

```python
# 1850000 VND is just for testing.
amount = int(input("Please enter your Amount (VND): ") or 1850000)

while amount < 50000:
    print()
    print("This amount is less than 50,000 VND! Please enter your amount again!")
    amount = int(input("Amount (VND):"))

while amount % 50000:
    print()
    print("It's not a multiple of 50,000! Please enter your amount again!")
    amount = int(input("Amount (VND):"))
```

```python
d = 50000
print(f"{d:,}")
```

```python
denominations = [500000, 200000, 100000, 50000]
remainder = amount

for d in denominations:
    print(f"{d:,} x", remainder // d)
    remainder = remainder % d
```
