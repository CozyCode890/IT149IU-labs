````python
def read_number(message: str, low, high):
    while True:
        text = input(message)
        if not text.isdigit():
            print("Please enter a number!")
            continue
        value = float(text)
        if not low <= value <= high:
            print("You are not a human!")
            continue
        return value

def read_choice(message, choices = ["walking", "running"]):
    while True:
        text = input(message).strip().lower()
        if text in choices:
            return text
        print(f"Just {choices[0]} or {choices[1]}!")

def input_helper():
    weight = read_number("Enter weight in kg ", 20, 300)
    minutes = int(read_number("Enter minutes exercised: ", 0, 20 * 60))
    exercise_type = read_choice("Enter exercise type (running/walking): ")
    return weight, minutes, exercise_type
````

````python
def output_helper(calorie):
    print(f"You burned approximately {calorie} calories")
````

````python
def calorie_calculate():
    w, m, t = input_helper()

    rate: float
    if t == "running":
        rate = 0.08
    else:
        rate = 0.05

    c = w * m * rate

    output_helper(c)
````

Simulate flipping a coin 100 times to see the probability of heads vs. tails, like in a fair game or betting scenario. Use the `random` module in a function to simulate the tosses and count heads and tails. The function should return the counts as a tuple. In the main program, unpack the tuple and display the percentages.

<font color="#C00000">**Extension:** add a parameter `flips=100` and run the simulation for 100, 10,000 and 1,000,000 flips. What happens to the percentages? (Law of large numbers.)</font>

### <font color="#0070C0">🔵 [REVISED] Task 4 · Level 2 ⭐⭐ (Basic) — Analysis → Algorithm</font>

- <font color="#0070C0">**① Understand the problem** — Input: number of flips (default 100). Output: (heads, tails) and their percentages.</font>
- <font color="#0070C0">**② Work an example by hand** — 4 flips give H, T, H, H ⇒ heads=3, tails=1 ⇒ 75% / 25%.</font>
- <font color="#0070C0">**③ Break it down** — a function counting heads, tails in a for loop → return a tuple → main: unpack, compute %</font>
- <font color="#0070C0">**④ Pseudo-code** (complete – trace it by hand before you code)</font>

<pre style="color:#0070C0;background:#EAF3FB;padding:8px 12px;border-left:4px solid #0070C0">FUNCTION toss(flips=100):
    heads &lt;- 0; tails &lt;- 0
    REPEAT flips TIMES:
        IF random() &lt; 0.5 THEN heads += 1 ELSE tails += 1
    RETURN heads, tails
MAIN: h, t &lt;- toss(); OUTPUT h/100*100, t/100*100</pre>

- <font color="#0070C0">**⑤ Self-check** — heads + tails must always equal the number of flips.</font>

````python
# ✍️ Task 4 – your code
# Write your code below, then run the cell (Shift+Enter).
import random

def toss(flips=100):
    heads = 0
    tails = 0
    for i in range(100):
        if random.random() < 0.5:
            heads += 1
        else:
            tails += 1
    return heads, tails

h, t = toss()
print(h, t)
````

````python
# ✍️ Task 7 – your code (basic + extended)
# Write your code below, then run the cell (Shift+Enter).
def total(*prices):
    return sum(p for p in prices)
````

````python
prices = []
p = float(input("Enter item price (0 to finish): "))

while p != 0:
    prices.append(p)
    p = float(input("Enter item price (0 to finish): "))
print(f"Total grocery bill: ${total(*prices)}")

# Enter item price (0 to finish): 150
# Enter item price (0 to finish): 89.5
# Enter item price (0 to finish): 45
# Enter item price (0 to finish): 0
# Total grocery bill: $284.50
````

- <font color="#0070C0">**④ Pseudo-code** (fill in the ___ blanks, then finish it in the ✍️ cell)</font>

<pre style="color:#0070C0;background:#EAF3FB;padding:8px 12px;border-left:4px solid #0070C0">total_games_played &lt;- 0
FUNCTION roll(n, sides=6):
    global total_games_played; total_games_played += ___
    rolls &lt;- [random 1..sides FOR each of n dice]
    RETURN min(rolls), ___, ___
MAIN: call 2–3 times, print total_games_played</pre>

- <font color="#0070C0">**⑤ Self-check** — After 3 calls the counter must equal 3.</font>

````python
# ✍️ Task 8 – your code
# Write your code below, then run the cell (Shift+Enter).
import random

total_games_played = 0
def roll(n, sides=6):
    global total_games_played;
    total_games_played += 1
    rolls = [random.randint(1, n) for _ in range(n)]
    return min(rolls)

print(roll(6))
print(roll(5))
print(total_games_played)
````
