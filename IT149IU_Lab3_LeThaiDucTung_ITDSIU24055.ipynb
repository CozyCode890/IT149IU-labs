✍️ **Task 3 – pseudocode**

````python
# ✍️ Task 3 – program code
# Write your code below, then run the cell (Shift+Enter).

def input_helper() -> tuple[int, int]:
    mothers = input("Enter the number of does in the rabbit colony (-1 to end): ").strip()

    if mothers == "" or mothers == "-1":
        return -1, 0

    if not mothers.isdigit():
        return 0, -1

    childs = input("Number of baby rabbits born in the past month: ").strip()

    if not childs.isdigit():
        return int(mothers), -1

    return int(mothers), int(childs)

def calculate_average(mothers, childs):
    if mothers == 0:
        return 0.0
    return childs / mothers

def show_sentinel(text: str):
    print(text)

def sentinel():
    total_mothers, total_childs = 0, 0
    while True:
        mothers, childs = input_helper()

        if mothers == -1:
            if total_mothers > 0:
                average_of_total = calculate_average(total_mothers, total_childs)
                show_sentinel(f"Total number of baby rabbits for each doe so far: {average_of_total}")
            else:
                show_sentinel("No rabbit data was entered")
            break

        if childs == -1:
            show_sentinel("Type again. Childs cannot be empty")
            continue

        total_mothers += mothers
        total_childs += childs
        average = calculate_average(mothers, childs)
        show_sentinel(f"On average {average} baby rabbits were born for each doe")

sentinel()
````

## Task 4: Challenge – Approximating the Mathematical Constant π

Write a script that computes the value of π from the following infinite series. Print a table that shows the value of π approximated by one term of this series, by two terms, by three terms, and so on. How many terms of this series do you have to use before you first get 3.14? 3.141? 3.1415? 3.14159?

**π = 4 − 4/3 + 4/5 − 4/7 + 4/9 − 4/11 + …**

> 📦 **Deliverables**
>
> 1. Provide the pseudocode.
> 2. Develop the corresponding program code.
> 3. Execute the program and present the results. Screenshots should clearly show the input and output.
>

> 💡 **Hint**
>
> <font color="#C00000">"First get 3.14" means the first approximation whose value *starts with* 3.14, i.e. `3.14 <= approx < 3.15` (you may also use `int(approx * 100) == 314`). Print only every 1000th row of the table to keep the output short, but keep checking every term.</font>
>

````python
# ✍️ Task 4 – program code
# Write your code below, then run the cell (Shift+Enter).

def check(i) -> int:
    if i % 2:
        return 1
    return -1

def element(i) -> float:
    denom = 2 * i - 1
    return 4 / denom

def input_helper() -> int:
    while True:
        text = input("Enter the number of loop iterations (positive integer): ").strip()
        if not text.isdigit():
            print("Please enter a whole number, like 10 or 1000.")
            continue

        n = int(text)
        if n == 0:
            print("The number must be greater than 0.")
            continue
        return n

def calculate_pi():
    n = input_helper()
    pi = 0
    for i in range(1, n + 1):
        sign = check(i)
        e = sign * element(i)
        pi += e
    return pi

calculate_pi()
````

<font color="#C00000">Read an odd number n (validate it with a `while` loop) and print a diamond of asterisks whose middle row has n stars. For n = 7:</font>

<font color="#C00000">*(added)*</font>
```text
   *
  ***
 *****
*******
 *****
  ***
   *
```

````python
# ✍️ Task 10 – your code
# Write your code below, then run the cell (Shift+Enter).
n = 7
middle = int((n - 1) / 2)
central = [middle, middle]

for y in range(n):
    for x in range(n):
        if x in range(central[0], central[1] + 1):
            print("*", end="")
        else:
            print(" ", end="")
    print()

    if y < middle:
        central[0] -= 1
        central[1] += 1
    else:
        central[0] += 1
        central[1] -= 1
````
