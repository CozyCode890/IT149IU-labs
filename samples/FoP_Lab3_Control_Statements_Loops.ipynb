# <font color="#C00000">3. Theory</font>

## <font color="#C00000">3.1 Counter-controlled vs. sentinel-controlled repetition</font>

- <font color="#C00000">**Counter-controlled** – the number of repetitions is known in advance → use `for i in range(n):`.</font>
- <font color="#C00000">**Sentinel-controlled** – the loop runs until a special value (the *sentinel*, e.g. `-1`) is entered; the number of repetitions is unknown → use `while`.</font>

<font color="#C00000">*(added)*</font>
```text
READ value
WHILE value != sentinel
    process value
    READ value            # read the next value at the END of the loop body
PRINT results
```

## <font color="#C00000">3.2 range()</font>

| <font color="#C00000">Call</font> | <font color="#C00000">Values</font> |
|---|---|
| <font color="#C00000">`range(5)`</font> | <font color="#C00000">0, 1, 2, 3, 4</font> |
| <font color="#C00000">`range(2, 6)`</font> | <font color="#C00000">2, 3, 4, 5</font> |
| <font color="#C00000">`range(0, 10, 3)`</font> | <font color="#C00000">0, 3, 6, 9</font> |
| <font color="#C00000">`range(10, 0, -2)`</font> | <font color="#C00000">10, 8, 6, 4, 2</font> |

## <font color="#C00000">3.3 Boolean operators and short-circuit evaluation</font>

| <font color="#C00000">a</font> | <font color="#C00000">b</font> | <font color="#C00000">`a and b`</font> | <font color="#C00000">`a or b`</font> | <font color="#C00000">`not a`</font> |
|---|---|---|---|---|
| <font color="#C00000">True</font> | <font color="#C00000">True</font> | <font color="#C00000">True</font> | <font color="#C00000">True</font> | <font color="#C00000">False</font> |
| <font color="#C00000">True</font> | <font color="#C00000">False</font> | <font color="#C00000">False</font> | <font color="#C00000">True</font> | <font color="#C00000">False</font> |
| <font color="#C00000">False</font> | <font color="#C00000">True</font> | <font color="#C00000">False</font> | <font color="#C00000">True</font> | <font color="#C00000">True</font> |
| <font color="#C00000">False</font> | <font color="#C00000">False</font> | <font color="#C00000">False</font> | <font color="#C00000">False</font> | <font color="#C00000">True</font> |

<font color="#C00000">`and` stops at the first `False`, `or` stops at the first `True` (**short-circuit**). Example: `if count != 0 and total / count > 5:` never divides by zero. Python also allows chained comparisons: `0 <= grade <= 100`.</font>

## <font color="#C00000">3.4 break and continue</font>

<font color="#C00000">`break` terminates the nearest enclosing loop; `continue` skips the rest of the current iteration. A loop's optional `else:` block runs only if the loop was **not** ended by `break`.</font>

## <font color="#C00000">3.5 Nested loops</font>

<font color="#C00000">The inner loop runs completely for **each** iteration of the outer loop. For patterns, the outer loop usually controls the *rows* and the inner loop the *columns*; `print(..., end='')` keeps output on the same line and a plain `print()` ends the row.</font>

## <font color="#C00000">3.6 Validating input</font>

````python
# [ADDED] worked example – run this cell
while True:
    text = input("Enter a positive integer: ")
    if text.isdigit() and int(text) > 0:
        n = int(text)
        break
    print("Invalid input – try again.")
print("You entered", n)
````

<font color="#C00000">For real numbers use `try: value = float(text)` / `except ValueError:` (exceptions are studied in detail later).</font>

# <font color="#C00000">4. Worked Examples</font>

<font color="#C00000">**Worked Example 1 – Class average with a sentinel.** Enter grades until `-1`; guard against division by zero.</font>

````python
# [ADDED] worked example – run this cell
total = 0
count = 0
grade = int(input("Enter grade, -1 to end: "))
while grade != -1:
    total += grade
    count += 1
    grade = int(input("Enter grade, -1 to end: "))

if count != 0:
    print(f"Class average is {total / count:.2f}")
else:
    print("No grades were entered")
````

<font color="#C00000">**Worked Example 2 – break and continue.** Sum the odd numbers from 1 upward, stop when the sum exceeds 50.</font>

````python
# [ADDED] worked example – run this cell
total = 0
for n in range(1, 100):
    if n % 2 == 0:
        continue          # skip even numbers
    total += n
    if total > 50:
        break             # leave the loop
print("last n =", n, "sum =", total)
````

<font color="#C00000">**Worked Example 3 – Nested loops: a right triangle of numbers.**</font>

````python
# [ADDED] worked example – run this cell
for row in range(1, 5):
    for col in range(1, row + 1):
        print(col, end=' ')
    print()
````
