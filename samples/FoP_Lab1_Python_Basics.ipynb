# 3. Theory

In Python, variables are assigned using `=`. Strings are enclosed in quotes. The `print()` function displays output and can accept multiple arguments separated by commas. By default, `print()` separates arguments with a space and ends with a newline.

## <font color="#C00000">3.1 Variables and basic data types</font>

<font color="#C00000">A **variable** is a name that refers to a value. Python is *dynamically typed*: the type belongs to the value, not to the name, and `type()` tells you the type.</font>

| <font color="#C00000">Type</font> | <font color="#C00000">Example</font> | <font color="#C00000">Meaning</font> |
|---|---|---|
| <font color="#C00000">`int`</font> | <font color="#C00000">`x = 7`</font> | <font color="#C00000">whole numbers (unlimited size)</font> |
| <font color="#C00000">`float`</font> | <font color="#C00000">`price = 7.5`</font> | <font color="#C00000">real numbers (about 15–16 significant digits)</font> |
| <font color="#C00000">`str`</font> | <font color="#C00000">`name = 'Ada'` or `"Ada"`</font> | <font color="#C00000">text (sequence of characters)</font> |
| <font color="#C00000">`bool`</font> | <font color="#C00000">`ok = True`</font> | <font color="#C00000">truth values `True` / `False`</font> |

<font color="#C00000">Names may contain letters, digits and `_`, cannot start with a digit and are case-sensitive (`total` ≠ `Total`). Use descriptive *snake_case* names such as `eggs_per_box`.</font>

````python
# [ADDED] worked example – run this cell
x = 2
y = 3.0
name = 'x'
print(type(x), type(y), type(name))   # <class 'int'> <class 'float'> <class 'str'>
print(x, name)                        # 2 x   -> variable value vs. the text 'x'
````

## <font color="#C00000">3.2 Arithmetic operators and precedence</font>

| <font color="#C00000">Operator</font> | <font color="#C00000">Name</font> | <font color="#C00000">Example</font> | <font color="#C00000">Result</font> |
|---|---|---|---|
| <font color="#C00000">`+`</font> | <font color="#C00000">addition</font> | <font color="#C00000">`7 + 2`</font> | <font color="#C00000">`9`</font> |
| <font color="#C00000">`-`</font> | <font color="#C00000">subtraction</font> | <font color="#C00000">`7 - 2`</font> | <font color="#C00000">`5`</font> |
| <font color="#C00000">`*`</font> | <font color="#C00000">multiplication</font> | <font color="#C00000">`7 * 2`</font> | <font color="#C00000">`14`</font> |
| <font color="#C00000">`/`</font> | <font color="#C00000">true division (always float)</font> | <font color="#C00000">`7 / 2`</font> | <font color="#C00000">`3.5`</font> |
| <font color="#C00000">`//`</font> | <font color="#C00000">floor division (rounds down toward −∞)</font> | <font color="#C00000">`7 // 2`, `-7 // 2`</font> | <font color="#C00000">`3`, `-4`</font> |
| <font color="#C00000">`%`</font> | <font color="#C00000">remainder (modulo)</font> | <font color="#C00000">`7 % 2`</font> | <font color="#C00000">`1`</font> |
| <font color="#C00000">`**`</font> | <font color="#C00000">exponentiation</font> | <font color="#C00000">`2 ** 10`</font> | <font color="#C00000">`1024`</font> |

<font color="#C00000">**Precedence (high → low):** `()` → `**` → unary `-` → `* / // %` → `+ -`. For example `-5 ** 2` is `-25` (the power is evaluated first) while `(-5) ** 2` is `25`. Use parentheses whenever in doubt.</font>

## <font color="#C00000">3.3 Input and type conversion</font>

<font color="#C00000">`input(prompt)` always returns a **string**. Convert it before doing arithmetic: `int('5')` → `5`, `float('2.5')` → `2.5`, `str(42)` → `'42'`. Converting text that is not a number raises a `ValueError`.</font>

## <font color="#C00000">3.4 Output: print() and f-strings</font>

<font color="#C00000">`print(*values, sep=' ', end='\n')` — `sep` is placed between the values, `end` after the last one. An **f-string** (`f"..."`) evaluates expressions inside `{ }` and can format them with a *format specifier* after a colon:</font>

| <font color="#C00000">Specifier</font> | <font color="#C00000">Example</font> | <font color="#C00000">Output</font> |
|---|---|---|
| <font color="#C00000">`:>8` right-align, width 8</font> | <font color="#C00000">`f"{42:>8}"`</font> | <font color="#C00000">`      42`</font> |
| <font color="#C00000">`:<8` left-align</font> | <font color="#C00000">`f"{'ab':<8}\|"`</font> | <font color="#C00000">`ab      \|`</font> |
| <font color="#C00000">`:^8` centre</font> | <font color="#C00000">`f"{'ab':^8}\|"`</font> | <font color="#C00000">`   ab   \|`</font> |
| <font color="#C00000">`:.2f` 2 decimals</font> | <font color="#C00000">`f"{3.14159:.2f}"`</font> | <font color="#C00000">`3.14`</font> |
| <font color="#C00000">`:,` thousands separator</font> | <font color="#C00000">`f"{1234567:,}"`</font> | <font color="#C00000">`1,234,567`</font> |
| <font color="#C00000">`:05d` zero padding</font> | <font color="#C00000">`f"{7:05d}"`</font> | <font color="#C00000">`00007`</font> |
| <font color="#C00000">`:.1%` percentage</font> | <font color="#C00000">`f"{0.256:.1%}"`</font> | <font color="#C00000">`25.6%`</font> |

## <font color="#C00000">3.5 Decisions with if / elif / else</font>

<font color="#C00000">Comparison operators `== != < <= > >=` produce `True` or `False`; combine conditions with `and`, `or`, `not`. The body of an `if` is the **indented** block (4 spaces) after the colon.</font>

````python
(5>2) or (10<=100): True
````

````python
1+3 == 10: False
````

True and True ==> True

````python
False and Any ===> False
````

False or False ===> False

True or Any ==> True

not True ==> False

not False ==> True

````python
# [ADDED] worked example – run this cell
n = 14
if n % 2 == 0:
    print(n, "is even")
elif n % 3 == 0:
    print(n, "is odd and divisible by 3")
else:
    print(n, "is odd")
````

<font color="#C00000">**Worked Example 2 – A formatted receipt (alignment and decimals).**</font>

````python
# [ADDED] worked example – run this cell
items = [("Notebook", 2, 12500), ("Pen", 5, 4000), ("Python book", 1, 185000)]
print(f"{'Item':<12}{'Qty':>5}{'Price':>10}{'Total':>12}")
grand = 0
for name, qty, price in items:
    total = qty * price
    grand += total
    print(f"{name:<12}{qty:>5}{price:>10,}{total:>12,}")
print(f"{'GRAND TOTAL':<27}{grand:>12,} VND")
````

## Exercise 4 – Eggs in boxes

Typically 6 eggs fit in one box. Write a script to calculate the number of boxes a farmer needs to store 28 eggs. The script will also determine how many eggs are placed in the last uncompleted box, and how many additional eggs are needed to fill this last box.

**Concept**

- Use floor division `//` to compute full boxes.
- Use modulo `%` to find the leftover eggs.
- If remainder > 0 → add 1 extra box.
- Eggs missing = `per_box - remainder`.

````python
eggs = 28
per_box = 6

boxes = eggs // per_box + (1 if eggs % per_box else 0)
last_box_fill = eggs % per_box if eggs % per_box else per_box
need_to_fill_last = per_box - (eggs % per_box) if eggs % per_box else 0

print("Total eggs:", eggs)
print("Boxes needed:", boxes)
print("Eggs in last box:", last_box_fill)
print("Eggs needed to fill last box:", need_to_fill_last)
````

## Exercise 8 – Name score with ord()

Each of a string's **characters** has an integer representation. The set of characters a computer uses together with the characters' integer representations is called that computer's **character set**. You can indicate a character value in a program by enclosing that character in quotes, as in `"A"`. To determine a character's integer value, call the built-in function `ord` (e.g. `ord('A')` → `65`). Tom and Jim want to play a game but are unable to choose who goes first. They decide that the person with the highest sum of the integer values of the characters of their name gets to play first. Write a script to determine who goes first.

**Concept:** `ord('A') = 65`, `ord('B') = 66`, etc. Sum `ord()` of all characters in a string to compute a name score, then compare the scores of Tom and Jim.

````python
def name_score(name):
    return sum(ord(c) for c in name)

tom_score = name_score("Tom")
jim_score = name_score("Jim")
print("Tom score:", tom_score)
print("Jim score:", jim_score)

if tom_score > jim_score:
    print("Tom goes first!")
elif jim_score > tom_score:
    print("Jim goes first!")
else:
    print("It's a tie!")
````

Tom score: 304
Jim score: 288
Tom goes first!

### Step-by-step guide

1. Input grades for three courses (e.g., Math, Physics, AI).

3. Compute the average grade using `sum(values) / len(values)`.

````python
courses = {}
courses['Math'] = int(input('Enter grade for Math: '))
courses['Physics'] = int(input('Enter grade for Physics: '))
courses['AI'] = int(input('Enter grade for AI: '))

avg = sum(courses.values()) / len(courses)
best = max(courses, key=courses.get)
worst = min(courses, key=courses.get)

print('Average grade:', round(avg, 2))
print('Highest grade:', best, courses[best])
print('Lowest grade:', worst, courses[worst])
````

**Concept:** the annual raise/cut is compounded multiplicatively; use the exponent operator `**` for powers, `(1 + p) ** n`; round the final wage to 2 decimals when printing.
